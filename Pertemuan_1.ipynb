```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pyplot as plt


# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcR3E35-ASd8fiRRqwhUb16yPkdU2PfM_N1zw7nr-BhISw&s=10"]
# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
  image = io.imread(url)                                #read image
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize image to half size
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)  #convert color to RGB
  final_frame = cv.hconcat((image, image_2))      #concatenate image
  cv2_imshow(final_frame)                         #show image
  print('\n')


#Melihat ukuran file image
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ",tinggi," x ",lebar)
cv2_imshow(image_2)


#Mengakses pixel dengan memberi garis diagonal
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#Membuat garis horizontal ditengah image
for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pyplot as plt


# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcR3E35-ASd8fiRRqwhUb16yPkdU2PfM_N1zw7nr-BhISw&s=10"]

#1.Mengakses pixel dengan memberi garis vertikal
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = image_2.copy()

# Membuat garis vertikal di tengah image
for y in range(tinggi):
    image_3[y, int(lebar / 2)] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)

#2.Mengakses pixel dengan memberi garis diagonal menyilang
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_4 = image_2.copy()

# Membuat garis diagonal dari kiri atas ke kanan bawah
for y in range(min(tinggi, lebar)):
    x = int(y * (lebar - 1)/(tinggi - 1))
    image_4[y, x] = [255, 255, 255]

# Membuat garis diagonal dari kanan atas ke kiri bawah
for y in range(tinggi):
    x = int((lebar - 1)-y*(lebar - 1)/(tinggi-1))
    image_4[y,x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_4))
cv2_imshow(final_frame)


#3. Membuat garis horizontal berwarna putih di tengah image
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = image_2.copy()

# Posisi garis
y_tengah = int(tinggi / 2)

# Panjang garis tertentu
panjang = 100

# Menentukan titik awal dan akhir
x_awal = int((lebar - panjang) / 2)
x_akhir = int((lebar + panjang) / 2)

# Menggambar garis horizontal
for x in range(x_awal, x_akhir):
    image_3[y_tengah, x] = [255, 255, 255]

# Menampilkan hasil
final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)


# 4. Membuat kotak menggunakan kumpulan pixel warna putih
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_4 = image_2.copy()

# Ukuran kotak
ukuran = 100

# Posisi kotak di tengah gambar
x_awal = 70
y_awal = 70

# Membuat kotak putih
for y in range(y_awal, y_awal + ukuran):
    for x in range(x_awal, x_awal + ukuran):
        image_4[y, x] = [255, 255, 255]

# Menampilkan hasil
final_frame = cv.hconcat((image_2, image_4))
cv2_imshow(final_frame)
```
